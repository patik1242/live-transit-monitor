```python
%sql
-- File: create_reconciliation_results.sql
CREATE TABLE IF NOT EXISTS dbr_dev.live_transit_monitor.reconciliation_results (
    check_name        STRING    NOT NULL COMMENT 'Which comparison was run',
    source_table      STRING             COMMENT 'Layer the keys were read from',
    target_table      STRING             COMMENT 'Layer the keys were compared against',
    source_keys       BIGINT             COMMENT 'Distinct business keys in the source',
    target_keys       BIGINT             COMMENT 'Distinct business keys in the target',
    missing_in_target BIGINT             COMMENT 'Keys present in source but absent from target',
    missing_in_source BIGINT             COMMENT 'Keys present in target but absent from source',
    checked_at        TIMESTAMP NOT NULL COMMENT 'When the check ran'
)
USING DELTA
COMMENT 'History of reconciliation checks between medallion layers.';
```
